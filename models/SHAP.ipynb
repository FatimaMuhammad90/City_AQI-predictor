```python
import shap
import pandas as pd
from catboost import CatBoostRegressor
import joblib
from models.preprocessing import preprocess_data
import time
import numpy as np
```

## For 24 hour, XGBoost

```python


try:
    xg_model = joblib.load("../models/xgboost_24h.pkl")
    print("XGBoost model loaded successfully!")
except FileNotFoundError:
    print("XGBoost model not loaded")


filepath = "../data/combined_air_weather_5_cities_features.csv"

X_train, X_test, y_train, y_test, encoder, train, test,  df_processed,test_cities,test_origins = preprocess_data(filepath,target_column="target_24h")
```

## Feature Importance for XGboost at 24 hour

```python
explainer = shap.TreeExplainer(xg_model)

shap_values = explainer.shap_values(X_test)

print("X_test shape:", X_test.shape)
print("SHAP shape:", shap_values.shape)

shap_importance = pd.DataFrame({
    "feature": X_test.columns,
    "importance": abs(shap_values).mean(axis=0)
})

shap_importance = shap_importance.sort_values(
    "importance",
    ascending=False
)
print(shap_importance)
```

### Positive/Negative Influence

```python
shap.summary_plot(
    shap_values,
    X_test
)
```
